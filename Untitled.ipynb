```python
class Student:
    def __init__(std, usn, name, place):
        std.usn=usn
        std.name=name
        std.place=place
    def show(std):
        print(f'USN of {std.usn} having the name {std.name} is from {std.place}')
    def update(std, usn, place):
        std.usn=usn
        std.place=place
```

```python
stud = Student(130, 'RMS', 'Mysuru')
stud.name
```

```python
stud.show()
```

```python
stud.update(19, 'Blr')
```

```python
stud.show()
```
